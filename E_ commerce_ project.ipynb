```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors
pio.templates.default="plotly_white"
```

```python
data=pd.read_csv("Amazon Sale Report.csv.zip",encoding='latin-1')
```

```python
data.head()
```

```python
data.describe()
```

```python
data.info()
```

#Converting Date columns

```python
data[' Date']=pd.to_datetime(data['Date'],format='%m-%d-%y')
```

```python
data.info()
```

```python
data.head()
```

```python
data.columns.tolist()
```

```python
data.columns=data.columns.str.strip()
```

```python
data.columns.tolist()
```

```python
data=data.loc[:,~data.columns.duplicated()]
```

```python
data['Date']=pd.to_datetime(data['Date'],errors='coerce')
```

```python
print(data['Date'].dtype)
```

```python
data['Order Month']=data['Date'].dt.month
data['Order Year']=data['Date'].dt.year
data['Order Day of Week']=data['Date'].dt.dayofweek
```

```python
data.head()
```

#What is the total revenue generated?

```python
total_revenue=data['Amount'].sum()
average_order_value=data['Amount'].mean()
```

```python
print("Total Revenue",total_revenue)
print("Average Order Value",average_order_value)
```

```python
revenue_data=pd.DataFrame({
    'Metric':['Total Revenue',
              'Average Order Value'],
    'Value':[total_revenue,average_order_value]})
fig=px.bar(
    revenue_data,
    x='Metric',
    y='Value',
    title='Revenue Performance',
    text='Value'
)
fig.update_traces(texttemplate='%{text:.2f}',textposition='outside')
fig.show()   
                 

              
    
```

 #How do sales vary month by month

```python
data['Month']=data['Date'].dt.month_name()
data['Month_Number']=data['Date'].dt.month
monthly_sales=data.groupby(
    ['Month_Number','Month'],
    as_index=False
)['Amount'].sum().sort_values('Month_Number')

monthly_sales

    
```

```python
fig=px.line(
    monthly_sales,
    x='Month',
    y='Amount',
    markers=True,
    title='Monthly Sales Trend'
)
fig.update_layout(
    xaxis_title='Month',
    yaxis_title='Total Sales'
)
fig.show()
```

3.# Which product category generates the highest revenue?

```python
category_sales = data.groupby(
    'Category',
    as_index=False
)['Amount'].sum().sort_values(
    'Amount',
    ascending=False
)

category_sales
```

```python
fig = px.bar(
    category_sales,
    x='Category',
    y='Amount',
    title='Revenue by Product Category',
    text='Amount'
)

fig.update_traces(
    texttemplate='%{text:.2s}',
    textposition='outside'
)

fig.show()
```

4.# Which are the top 10 best-selling products?

```python
top_products = data.groupby(
    'SKU',
    as_index=False
)['Qty'].sum().sort_values(
    'Qty',
    ascending=False
).head(10)

top_products
```

```python
fig = px.bar(
    top_products,
    x='Qty',
    y='SKU',
    orientation='h',
    title='Top 10 Best-Selling Products',
    text='Qty'
)

fig.update_layout(
    yaxis={'categoryorder': 'total ascending'}
)

fig.show()
```

7.# What is the distribution of order statuses?

```python
status_counts = data['Status'].value_counts().reset_index()

status_counts.columns = ['Status', 'Count']

status_counts
```

```python
fig = px.pie(
    status_counts,
    names='Status',
    values='Count',
    title='Order Status Distribution',
    hole=0.3
)

fig.show()
```


8.# How is the courier delivery performance?

10.# How does B2B sales compare with B2C sales?

```python
b2b_sales = data.groupby(
    'B2B',
    as_index=False
)['Amount'].sum()

b2b_sales['Customer Type'] = b2b_sales['B2B'].map({
    True: 'B2B',
    False: 'B2C'
})

b2b_sales
```

```python
fig = px.bar(
    b2b_sales,
    x='Customer Type',
    y='Amount',
    title='B2B vs B2C Revenue',
    text='Amount'
)

fig.show()
```
